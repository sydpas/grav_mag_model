```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

creating a pandas dataframe

```python
dt = pd.read_csv('..\\data\\ecm1\\Hcc_ECM1.csv', header=None, skiprows=1, names=['num', 'long', 'lat', 'hcc'])
print(dt)
```

```python
dt_dt = dt.drop(columns='num')
```

```python
df = dt_dt[dt_dt['long'].between(-132, -124) & dt_dt['lat'].between(39, 53)]

df.to_csv('..\\created_data\\trimmed_Hcc_ECM1.csv', index=False)
```

dataframe is a list with one row per point

pivot rearranges the dataframe into a grid-like spreadsheet
- index: values are row labels
- columns: values are column labels
- values: these values fill the cells

> ### example!
>
> dataframe shape: (64800, 3) and type: <class 'pandas.DataFrame'>
>
> pivot grid shape: (4, 4) and type: <class 'pandas.DataFrame'>

```python
df_grid = df.pivot(index='lat', columns='long', values='hcc')
```

```python
lat = df_grid.index.values  
long = df_grid.columns.values
hcc = df_grid.values
```

```python
fig, ax = plt.subplots(figsize=(8,6))

pcm = ax.pcolormesh(long, lat, hcc, cmap='rainbow', shading='auto')

fig.colorbar(pcm, label='Thickness [km]')

ax.set_title('Crystalline Crust Thickness')
ax.set_xlabel('Longitude')
ax.set_ylabel('Latitude')

plt.show()
```
