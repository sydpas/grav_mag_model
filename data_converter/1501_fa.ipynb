```python
rows = []
for l in open('..\\data\\RC1501.a77'):
    if not l.startswith('5'):
        continue
    l = l.rstrip('\r\n')
    faa = l[103:108]              # positions 104-108
    if faa == '+9999':            # fill value
        continue
    rows.append({
        'year': int(l[12:16]), 'month': int(l[16:18]), 'day': int(l[18:20]),
        'hour': int(l[20:22]), 'min': int(l[22:27]) / 1000,
        'lat': int(l[27:35]) / 1e5,
        'lon': int(l[35:44]) / 1e5,
        'FA': int(faa) / 10,
    })
```

```python
df = pd.DataFrame(rows)

print(df)
```

```python
df[['lon', 'lat', 'FA']].to_csv('..\\created_data\\seismic\\RC1501_FA.csv', index=False)
```
